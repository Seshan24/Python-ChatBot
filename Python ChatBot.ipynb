```python
# import groq
from groq import Groq
client = Groq()


messages =[]

while True:
  message = input("Your Message : ")
# append history
  messages.append( {
          "role":"user",
          "content":message
      })
  # call LLM
  response = client.chat.completions.create(
      model="openai/gpt-oss-120b",
      messages=messages
  )

  ai_response = response.choices[0].message.content

  messages.append( {
          "role":"system",
          "content":ai_response
      })

  print("Bot Reply : ",ai_response)
```
